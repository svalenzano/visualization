```python
import seaborn as sns 
import numpy as np 
import maplotlib.pyplot as plt
import pandas as pd
import scipy 
import PIL
import requests 
```

```python
tips = sns.load_dataset('tips')
print(tips)

sns.lineplot(data = tips, 
x = 'total_bill',
y = 'tip')
```

```python
sns.set_style('whitegrid')
```

```python
tipgraph = sns.lineplot(data=tips,
x='total_bill',
y='tip')

tipgraph.set(title='Total Bill vs Tip Amount',
xlabel='Total Bill ($)', ylabel='Tip Amount ($)')
```

```python
fig = sns.subplots(figsize=(7,3))
tipgraph = sns.lineplot(data=tips,
x='total_bill',
y='tip', 
color= 'hotpink',
linestyle = "--",
marker = "o",
markerfacecolor = 'indigo')
```

```python
tipgraph = sns.scatterplot(data=tips, x = 'total_bill',
                        y = 'tip', hue = 'day', style = 'time', palette = ['purple', 'hotpink', 'deepskyblue'] )

tipgraph.set(title='Total Bill vs Tip Amount',
xlabel='Total Bill ($)', ylabel='Tip Amount ($)')
```
