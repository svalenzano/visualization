```python
# first make our sample data 
x = np.array(["luffy", "zoro", "nami", "usopp", "sanji"])
y1 = np.array([100, 180, 240, 99, 220])
y2 = np.array([90, 160, 270, 120, 200])

# define our figure and axes (just do this one time)
fig, ax = plt.subplots(figsize=(7,3))

# now call both bar and plot elements to do the same axes 
ax.bar(x,y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
# add error information 
y2_sd = np.std(y2) 

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = "red")

ax.errorbar(x, y2, yerr = y2_sd, fmt = "none")

```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, y2, yerr = y2_sd, fmt ="none", ecolor = "indigo", elinewidth=4, capsize =6 )
```

```python
from PIL import Image 
import requests 
from io import BytesI0 

response = requests.get("https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_d_Luffy.png")
image_file = BytesI0(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = "red") 

ax_image = fig.add_axes([0.6, 0.5, 0.25, 0.4]) # [left, bottom, width, height]
ax_image.imshow(image)
ax_image.axis('off')  # Hide axes for the image
```
